```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
# Aman N Shah 22MID0250
# Load the dataset
file_path = "dataset for preprocessing.csv"  # Replace with your file path
df = pd.read_csv(file_path)

# Display the first five rows
print("First five rows of the dataset:")
print(df.head())

# Print summary statistics and data types
print("\nSummary statistics:")
print(df.describe(include="all"))
print("\nData types:")
print(df.dtypes)

# Identify columns with missing values
missing_values = df.isnull().sum()
print("\nColumns with missing values:")
print(missing_values)

# Calculate the percentage of missing values in each column
missing_percentage = (missing_values / len(df)) * 100
print("\nPercentage of missing values in each column:")
print(missing_percentage)

# Handling missing values
print("\nHandling missing values:")
# Replace missing "Name" with "Unknown"
df['Name'] = df['Name'].fillna("Unknown")
# Replace missing "Age" with median age
df['Age'] = df['Age'].fillna(df['Age'].median())
# Replace missing "Salary" with mean salary
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
# Replace "unknown" in "Joining_Date" with NaN
df['Joining_Date'] = df['Joining_Date'].replace("unknown", np.nan)
# Replace missing "Department" with "Unknown"
df['Department'] = df['Department'].fillna("Unknown")

# Check for duplicate rows
duplicates = df.duplicated()
print(f"\nNumber of duplicate rows: {duplicates.sum()}")

# Remove duplicate rows if any
df = df.drop_duplicates()
print("\nDuplicate rows removed.")

# Identify inconsistencies in Joining_Date
# Standardize the date format to YYYY-MM-DD
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'], errors='coerce', dayfirst=True)

# Identify invalid email addresses and replace them with NaN
df['Email'] = df['Email'].apply(lambda x: np.nan if x == "unknown" or '@' not in str(x) else x)

# Normalize the Salary column using Min-Max Scaling
scaler = MinMaxScaler()
df['Salary_Normalized'] = scaler.fit_transform(df[['Salary']])

# Encode the Department column using Label Encoding
encoder = LabelEncoder()
df['Department_Encoded'] = encoder.fit_transform(df['Department'])

# Display the cleaned dataset
print("\nCleaned dataset:")
print(df.head())

# Save the processed dataset (optional)
df.to_csv("processed_dataset.csv", index=False)
```
